```python
# Conferindo os cinco sinais - A (amplitude), phi (fase) e P (periodo)
x, n = SIGNALgenerate(30, 'delta',  k=4, A=2.0)
SIGNALplot(x, n, 'delta   2 d[n-4]')

x, n = SIGNALgenerate(30, 'step',   k=4, A=1.5)
SIGNALplot(x, n, 'step   1.5 u[n-4]')

x, n = SIGNALgenerate(30, 'exp',    a=0.8, A=3.0)
SIGNALplot(x, n, 'exp   3 (0.8)^n u[n]')

x, n = SIGNALgenerate(40, 'osc',    P=15, phi=np.pi/2, A=0.5)
SIGNALplot(x, n, 'osc   0.5 sin(2 pi n / 15 + pi/2)')

x, n = SIGNALgenerate(50, 'random', A=2.0)
SIGNALplot(x, n, 'random gaussiano, A = 2')
```

```python
# Atraso: o 2o argumento e o numero de amostras (inteiro), nao o vetor n
x, n = SIGNALgenerate(40, 'osc', period=10)
xd   = SIGNALdelay(x, 5)

SIGNALplot(x,  n, 'osc original')
SIGNALplot(xd, n, 'osc atrasada em 5 amostras')

# a prova: atrasar o delta em 7 == gerar o delta ja com k=7
import numpy as np
print('delay(delta, 7) == generate(delta, k=7) ?',
      np.array_equal(SIGNALdelay(SIGNALgenerate(20, 'delta')[0], 7),
                     SIGNALgenerate(20, 'delta', k=7)[0]))
```
